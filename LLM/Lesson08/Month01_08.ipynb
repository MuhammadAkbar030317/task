## Embedding

```python
model = AutoModel.from_pretrained("gpt2")
```

```python
print(model)
```

```python
print(model.wte)
print(model.wpe)
```

```python
#Demak 50257 ta token yani shuncha vocabulary ni token ID bilan taniy oladi ex: ID:0 --> cat ...ID:50256-->bye.  va 768 dimention yani 1 ta token vector ichida 768 ta son saqlaydi
# 1024 esa model 1024 ta position uchun embedding borligini anglatadi
```

```python
print(model.wte.weight.shape)
```

```python
print(model.wpe.weight[4])
```

```python
ids=tokenizer.encode(sentence)
ids
```

```python
inputs = tokenizer(sentence, return_tensors="pt")

embeddings = model.wte(inputs["input_ids"])

print(embeddings.shape)
```

```python
print(embeddings[0][0][:10])
# [0]- bu batch raqami
# [0]- bu 1-turgan so'zni computer shunaqa ko'ryabdi
# [:10] - esa vector ichiga 10 ta son kiryabdi.
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("EleutherAI/gpt-neo-125m")
```

```python
model1 = AutoModel.from_pretrained("EleutherAI/gpt-neo-125m")
```

```python
print(model1.wte)
print(model1.wpe)
```

```python
print(model1.wte.weight.shape)
```

```python
print(model1.wpe.weight[8])
```

```python
text = "I am a student at seojeong university"

tokens = tokenizer.tokenize(text)
ids = tokenizer.encode(text)

print("Tokens:", tokens)
print("Token IDs:", ids)
# seojeong inglizcha so'z emasligi uchun mana shunaqa subwords qilib tashlayabdi
```

```python
inputs = tokenizer(text, return_tensors="pt")

embeddings = model1.wte(inputs["input_ids"])

print(embeddings.shape)
```

```python
print(embeddings[0][2][:10])
```
