### NewGELU vs SiLU

```python
# NewGELU GPT2 da ishlatilingan va SiLu SmolLM2-360M da ikkisi bir xil qiymat kirsa qanday hisoblashini misol bilan ko'rsatishga harakat qildim.
```

```python
import torch
import torch.nn.functional as f

x = torch.tensor([-3., -1., 0., 1., 2., 3.])

new_gelu = 0.5 * x * (
    1 + torch.tanh(
        (2 / torch.pi) ** 0.5 * (x + 0.044715 * x ** 3)
    )
)

silu = f.silu(x)

print("Input:   ", x)
print("NewGELU: ", new_gelu)
print("SiLU:    ", silu)
```
